### List
- an collection of data
- ordered: every element has an index
- mutable: every element can be changed
- heterogenous: multiple types of data it can save

```python
fruits = ['apple', 'banana','cherry','dragon fruit', 'e', 'frooti']
```

```python
# accessing elements in a list
fruits[4]

# updating list
fruits[4] = 'eliachi'
```

```python
for fruit in fruits:
  print(fruit)
```

```python
# if you want position/index too: we need enumerate - OPTIONAL
for fruit in enumerate(fruits):
  print(fruit)
```

```python
fruits.append('mango')
print(fruits)
```

```python
last_element = fruits.pop() # pop returns the last element and removes it from our list
last_element
```

## Set - unordered unique set of items
- just a collection of **unique** items
- does not allow repetition
- unordered - 'set' object is not subscriptable
-

```python
set_a = {1,2,3,4,5,6,2,3,4,4,4,4,4}
print(set_a)

set_b = {4,5,6,7,8,9,10}
print(set_b)

# union
set_a | set_b # elements of both sets

# intersection
set_a & set_b # common elements
```

## Dictionary
- are structures that stores data in key value pairs
- key:value
- keys cannot be changed, values you can change
-
- `{key : value, key2: value}`

```python
# students roll number dictionary
students = {
    'richa': 34,
    'priyank': 35,
    'abhimanyu':36,
    'sonu':37,
    'richa': 35 # duplicate keys are not allowed - only the value will be updated
}
students
```

```python
# change values of keys that already exist - Values can be updated
students['abhimanyu'] = 66

# adding a new key value pair -
students['Idrees'] = 355

# deleting data from dictionary
del students['abhimanyu']

print(students)
```

```python
dict_a = {
    'a': {20},
    'b':78,
    'c':'hieee',
    'a':{60}
  }

# accessing an element - just search for the key
dict_a['c']

# change values of keys that already exist - Values can be updated
dict_a['c'] = {55}



print(dict_a)
```
