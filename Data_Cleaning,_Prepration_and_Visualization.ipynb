## Check for Missing values and drop them
- Imputation Methods
-

```python
from numpy import column_stack
df.isnull().sum() # pandas ka .sum

df_clean = df.drop(columns= ['deck', 'age'])

df_clean = df_clean.dropna()
df_clean.isnull().sum()

df_clean.info()
```

## Check for Duplicate values and drop them

```python
#df_clean
```

```python
df_clean.duplicated().sum()

df_clean = df_clean.drop_duplicates()
# we have cleaned any duplicates
```

```python
df_clean.duplicated().sum()

df_clean.info() # 514 rows of data
```
