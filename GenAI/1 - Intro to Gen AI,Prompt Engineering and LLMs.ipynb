## 3. What is an LLM and how does it work?

### What is a Large Language Model?

- **Large:** models often have many learned numerical settings and are trained on extensive data; “large” is relative, not an official minimum size.
- **Language:** they are built to process language, though modern models may also accept images, audio or video.
- **Model:** a learned computational system-not a person or a database of perfectly verified facts.

**An LLM is not the same thing as the chatbot interface.** 
- For example, *GPT* Generative Pretrained Transformer, is a name for model families, while *ChatGPT* is an application that can use those models for answering user questions
- 

**Training vs inference**

| Stage | What happens? | Easy analogy |
|---|---|---|
| Pre-training | A model learns statistical patterns from large datasets | Extensive study and practice |
| Post-training / Finetuning | Model behavior is shaped to follow requests and preferences | Practising how to answer questions clearly |
| Inference | An already trained model processes your request and generates output | Answering a new question |

The analogy is limited: machine learning is not literally human understanding.

### What happens when you type a prompt?

```text
**PROMPT** = Your question + any attached context
     ↓
Tokenised = Text represented as tokens = smaller chunks of big text
     ↓
Eembeddings = aka word vectors = Tokens given numerical representations - because LLMs understand numbers only
     ↓
Transformer = process relationships between those numbers = i.e. relations between words
     ↓
Predictions = Model scores possible next tokens = Probabilities of multiple next words = highest probability word is taken
     ↓
A token is selected → then another → ...
     ↓
Answer shown to you
```

**Three important points:**

1. A token is **not always one word**-it may be a word, a word fragment or punctuation.
2. A model generates a **probability distribution** over possible next tokens; it need not choose the single highest-probability token every time.
3. In many GPT-style models, input tokens can be processed together during prompt processing, but generated tokens are produced sequentially.

## **Transformer Architecture**

- In Deep Learning: CNNs were used for Vision, RNNs were used for NLP (Natural Language Processing)
- In 2017 Transformers replaced RNNs (Recrurrant Neural Network) by using a mechanism called **self-attention**

- Simplified main Components:
    1. **Input Embedding**: Convert words to vectors = Tokens convert to embeddings, add positional encodings
    2. **Self-Attention**: Assign weight to words = embeddings into Queries, Keys, and Values so each token attends to relevant tokens across a sequence
    3. **Feed-Forward Layer**: Compute token relevance = position-wise non-linear transformations to enrich token representations
    4. **Output**: Sequence of predicted tokens = representations pass through a projection head to produce next-token logits

<details><summary> <b>How did we go from RNN to Transformers?** [OPTIONAL]</b></summary>

- **RNNs:**
    - 
    - Imagine the case of, Sentence Understanding
    - Input Sentence:“The cat sat on the mat.”
    - **RNN Approach**: 
        - sequential data one step at a time 
        - while maintaining a hidden state 
        - that acts as a *running memory* 
        - of earlier tokens
    - Step-by-Step Processing:
        - RNN reads one word at a time:
        - Step 1: “The”
        - Step 2: “cat” (uses a hidden state from Step 1)
        - Step 3: “sat” (uses a hidden state from Step 2)
        - …and so on
    - Sequential context, of the all previous words = in terms of numbers obviously, because machine understand numbers only
        - all words are treated as numbers - aka word vectors - aka word embeddings 
            - which actually are a list of numbers - not just a single number
    - imagine an RNN saves all the previous word/token as a number (hidden state) 
        - acually its a vector, i.e. it can be more than one numbers, but let's not go there rn
    - when it sees the next number, it adds the next one to it
        - it sees the next number, it adds it, and so on...
    - So at each state, 
        - some weight/context from all previous words is kept 
        - and that context is just a number - a vector (called hidden state) 
        - so this number carries information of all previous words 
            - especially the immediate previous words
    
    - NOW
    - Over long sentences, relationships get more and more diluted with addition of more numbers
    - so the words at the start of sentence, are kind of way too far 
    - **Forgetting** 
        - The word 'cat' may be forgotten by the time the sentence reaches "the mat"


- This forgetting problem was solved with LSTMs
- **LSTMs:**
    - 
    - We added a Dedicated "Long-Term Memory" Highway (Cell State) 
        - aka (LSTM Long Short-Term Memory)
    - Alongside the short-term working memory, 
        - a separate state that saves only the important words
        - Like it only adds 'cat' to the cell state, skips filler words like 'the'
        - this means, important words will face lesser dilution
    - so short-term memory (hidden state) > saves context of all previous words
    - and a long term memory (cell state) > saves the context across the steps > working memory

- But the whole process was still **very slow**
    - 
    - especially because everything happened sequentially
        - Step 1: “The” $\rightarrow$ Step 2: “cat” $\rightarrow$ Step 3: “sat” $\rightarrow$ Step 4: “on” $\rightarrow$ Step 5: “the” $\rightarrow$ Step 6: “mat”
    - imagine waiting for a machine to go through every word of a sentence one by one
    - understand it > figure out relationships > predit next words ??!!!
    - So training was super slow too
        - Cannot process all 6 words at once on a GPU because Step 6 must wait for Steps 1–5
    - So, we moved to >

- **Transformers**
    - 
    - Parallel Processing (All Words at Once)
        - Instead of reading word-by-word, the Transformer feeds all 6 word vectors `["The", "cat", "sat", "on", "the", "mat"]` into the model simultaneously -
            - now parallel processing can be used i.e. all words at once 
        - since everything goes in at once, there is no position order 
            - so it adds Positional Encodings (order tags) to each vector 
            - so that the model still knows "cat" came before "mat"
    - Self-Attention (Direct Word-to-Word Connections)
        - Every word looks directly at every other word in the sentence in a single step
            - no passing notes down a chain
        - When processing "mat", self-attention connects it directly to "sat" and "cat" with high attention weights
            - while assigning low weights to filler words like "the"
        - Zero forgetting across distance: Whether "cat" and "mat" are 4 words apart or 4,000 words apart, the connection path is always 1 step

### Context window vs memory

**Context window:** Roughly, how many tokens a particular model request can handle together (prompt, history, retrieved content and often output, depending on the provider's definitions).

**Knowledge cutoff:** The latest date represented by a model's training knowledge; *not* a guarantee that everything before it is known.

**Chat memory:** A product feature that may recall selected information between chats. It is **not** identical to the model's context window.

**Quick question:** Would a model with a one-million-token context window automatically remember everything forever? **No.**

## 4. Evolution of LLMs and current models

### A useful historical timeline

| Period | Representative development | Why it matters |
|---|---|---|
| 2017 | Transformer architecture | Foundation for many modern language models |
| 2018–2020 | BERT, GPT-2, T5, GPT-3 | Major progress in text understanding and generation |
| 2022–2023 | ChatGPT, GPT-4, Claude, Bard | Conversational AI became mainstream; multimodal features expanded |
| 2024–2025 | Better multimodal systems, reasoning models, more open-weight models | More capable use of images, code and longer contexts |
| 2026 | New generations of GPT, Gemini, Claude, DeepSeek and Grok | Rapid changes in reasoning, context, tool use and modality support |

*Historical note:* Google's Bard was renamed **Gemini** in February 2024. The timeline describes broad trends, not exclusive periods.

### How to read a model specification table

| Field | What it tells you | What it **doesn't** tell you |
|---|---|---|
| **Context window** | Maximum supported token context for the specified model/API configuration | It does not guarantee perfect recall or accurate reasoning over every token. |
| **Parameters** | Learned numerical settings in the model | More parameters do not automatically mean better answers. |
| **Active parameters** | Approximate parameters involved in a token's processing in some architectures (e.g., MoE) | Not directly comparable with total parameters or model quality. |
| **Multimodal inputs** | Which input types a model accepts (text/image/audio/video) | Does not mean every app exposes every modality. |
| **Reasoning** | Capabilities or modes designed for harder multi-step tasks | Does not guarantee correctness. |
| **Open-weight** | Model weights are distributed under a licence | It does not necessarily mean open-source training data or unrestricted use. |
| **API vs app** | Model technical limits vs consumer application features | A chatbot subscription does not automatically offer the API's full limits. |

## 6. **Prompt engineering: concepts and techniques**

### What is prompt engineering?

**Prompt:** The instructions, questions, examples or other content you give an AI model/application.

**Prompt engineering:** Deliberately shaping the task and supplied context to make an AI system's output more useful, consistent or verifiable.


**Structure prompts like this:**

> **Task + Context + Constraints + Format**  

**Poor prompt:** “Explain this data.”

**Better prompt:** “I know Excel and basic SQL but I'm new to AI. In 5 short bullet points, explain how Generative AI can help a Data Analyst. Include one limitation and avoid jargon.”

- The second prompt provides a **goal**, **audience**, **format** and **constraints**. 
- A longer prompt is **not automatically** a better prompt: include only useful details.

- Other examples:
    - Summarize the pros and cons of online learning [Task] for high school teachers [Context] in 5 bullet points [Format]
    - Explain the impact of social media on teenage mental health [Task] for parents of teenagers [Context] in 4 short bullet points [Format]

#### Domains Prompt Engineering is used in
Pretty much every field out there:
- Education: Create study notes, quiz questions, lesson plans
- Content Creation: Write blog posts, design ideas, marketing content
- Research: Summarize articles, gather topic insights, generate citation formats
- Business: Draft emails, reports, analyze data This makes it a versatile skill for multiple career paths.

### Techniques you need to recognize

| Technique | What it means | When it can help |
|---|---|---|
| **Zero-shot** | Ask without providing task examples | Common, familiar tasks |
| **One-shot** | Include one example | Showing a particular output style |
| **Few-shot** | Include multiple examples | Specialized labels, consistency, formatting |
| **Role / audience** | Specify perspective or audience | Tailoring tone and depth (not guaranteeing expertise) |
| **Act As** | Assigns a domain identity, perspective, or standard (e.g., "Act as a senior database architect...") | effective for boosting raw reasoning accuracy on modern models
| **Instruction-based** | State explicitly what to do | Most tasks |
| **Question-chaining** | Split work into successive questions | Multi-stage work and clarification |
| **Chain-of-thought (historical term)** | Prompting for intermediate reasoning | Know the concept; modern reasoning models may not need “think step by step.” Ask for concise justifications and checks instead. |
| **Iterative refinement** | Improve instructions after observing errors | Almost every practical AI workflow |

**Important distinction:** 
- Asking the model to act *as an expert* does not make its output certified expert advice. 
- Asking for steps is not a substitute for verifying the result.

### One-shot is not few-shot

**One-shot (one example):**

```text
"Your delivery was late" → Complaint
Classify: "The product never arrived"
```

**Few-shot (three examples):**

```text
"Please refund my payment" → Refund request
"The screen is broken" → Complaint
"Where is my package?" → General inquiry
Classify: "I want my money back"
```

**Ask students:** What pattern are the examples teaching? What happens if you include contradictory examples?

#### Prompt Frameworks
Framework |	Structure | Best For
---|--|--
CO-STAR	| Context, Objective, Style, Tone, Audience, Response format |	Content generation, training materials, tailored explanations
RISEN |	Role, Instructions, Steps, End goal, Narrowing (constraints) |	Multi-step workflows, SOPs, analytical breakdowns
RTF |	Role, Task, Format |	Quick everyday prompts and clean structured outputs
RACE |	Role, Action, Context, Expectation |	Problem-solving, debugging, decision framing
ReAct |	Reasoning + Acting (thought -> tool/action -> observation loop) |	Agentic tasks, API/tool calls, multi-hop research
Chain-of-Thought (CoT) |	Step-by-step intermediate reasoning before final answer |	Multi-step logic, math, quantitative analysis


These matter much less on modern frontier and reasoning models than they used to. 

What remains high-leverage:
 - Context & specification engineering: Defining clear boundaries, schemas, edge cases, acceptance criteria, and ground truth.
 - Tool & agent orchestration: Designing clear tool contracts, fallback rules, and verification steps in agent loops.
 - Evaluation & iteration: Testing prompts against edge cases and failure modes rather than tuning wording by guesswork


A well-written skill (SKILL.md or agent instruction module) benefits from elements borrowed from these frameworks:
- Role & Scope (from RTF / RISEN / RACE): Establishes when the skill applies, what it owns, and what it explicitly avoids.
- Ordered Steps & Constraints (from RISEN): Defines execution order, safety guardrails, and output formatting rules.
- Reasoning + Action Loops (from ReAct): Guides how an agent checks state, calls tools or scripts, inspects results, and recovers from errors


### Designing Better Prompts:
Good prompts are built on these principles:

1. Principle 1 : **Clarity and Specificity**
Be clear and avoid vague terms. Mention exactly what you want.
- Example:
    - Weak: “Write about exercise.”
    - Strong: “List 3 benefits of regular exercise for college students.”


2. Principle 2: **Contextual Information**
Add background or user-specific info. It makes the AI’s output more relevant.
- Example: 
    - “As a biology professor, explain DNA replication to undergraduates.”

3. Principle 3: **Format Instructions**
Tell the AI how you want the response: bullets, tables, paragraphs. 
- Example: 
    - “Provide a comparison between solar and wind energy in a table format.”

4. Principle 4: **Progressive Refinement**
Start with a basic prompt, then refine based on the response.
- Example: 
    - Ask for an outline → Ask to expand sections → Finalize draft.

### Prompting example

Use the same AI assistant and start a new chat for each version where practical:

**Version A:**
```text
Explain SQL joins.
```

**Version B:**
```text
Explain SQL joins to someone who knows Excel but is new to databases.
```

**Version C:**
```text
Explain INNER JOIN and LEFT JOIN to a beginner who knows Excel.
Use a tiny Customers and Orders example with 3 rows each.
Show what happens to customers without orders.
Return: (1) a simple explanation, (2) two small result tables,
(3) one common mistake. Do not assume columns not specified.
```

**Ask:** Which modifications had a measurable effect? Did the model actually meet every requirement?

### Prompt quality checklist

Before sending, look for:

- [ ] The task is explicit.
- [ ] Relevant audience or business context is included.
- [ ] Necessary input data or examples are supplied.
- [ ] Output format and length are clear.
- [ ] Uncertainty is allowed (“say what is unknown”).
- [ ] Privacy-sensitive information has been removed.
- [ ] There is a way to verify the result.

**Prompting ≠ magically unlocking facts.** 
- The assistant still needs relevant information and, for certain tasks, access to appropriate tools.

#### How does AI understand Prompts
Couple of terms you will hear around. There is a lot of science behind each one, but we don't need to go too deep into understanding them
- Tokenization : The input prompt is broken down into smaller parts called tokens (words or subwords).
- Embedding : Each token is converted into a vector (a list of numbers) that represents its meaning in context.
- Context Analysis (Transformer Architecture) : AI uses a Transformer model that understands the order and relationship between words.
- Attention Mechanism : The model identifies which words are most important in the sentence using self-attention.
- Pattern Matching : Based on its training on huge amounts of text, the model recognizes patterns in your input (e.g., commands, questions, summaries).
- Understanding Intent : AI doesn't "understand" like humans, but it predicts your intent (e.g., are you asking a question or giving a task?).
- Response Generation : It generates a response one word (token) at a time, by predicting the most likely next word based on the prompt.
- Iteration : This process continues until the model decides the response is complete (based on probabilities).

## 8. Hallucinations, limitations and responsible AI

### Common limitations and how to address them

| Risk | What can go wrong? | Better practice |
|---|---|---|
| **Hallucination** | AI confidently invents a fact, citation or number | Check reliable primary sources and underlying data |
| **Faulty calculations** | AI writes convincing but incorrect numeric summaries | Recalculate with Excel, SQL, Python or a trusted calculator |
| **Missing context** | AI guesses what your company policy says | Supply authorized reference material; ask for evidence |
| **Privacy** | Sensitive customer or company information is disclosed | Use approved tools, anonymize and follow policies |
| **Bias** | Output reflects skewed examples or stereotypes | Review fairness and possible missing perspectives |
| **Outdated facts** | A model relies on older training knowledge | Use current sources or search tools; verify dates |
| **Prompt injection** | Instructions hidden in outside material try to override the task | Treat external text as data, not instructions |
| **Overreliance** | A learner accepts output without understanding it | Require a human check of important results |

### Hallucination Exercise

> **First, ask the model:**

```text
Our company has 12% lower revenue this quarter.
Explain exactly which products, cities and customers caused the decline.
```

**Observe:** 
- Can the assistant know those causes from this statement alone? 
- **No.** The question provides only an overall decline; the requested breakdown is unknown.

> **Rewrite the prompt:**

```text
We know only that overall revenue declined by 12% this quarter.
List which additional datasets and comparisons would be needed to
investigate possible causes. Do not invent product, city or customer figures.
Separate known facts from hypotheses.
```

**What changed?**
- The revised task asks for a plan of evidence, not unsupported conclusions.

**Here is a question about data privacy:** 
- Would you upload a real payroll spreadsheet or customer phone-number list to an unapproved AI account? **No.**

## 9. Recap and knowledge check

### Knowledge check

1. Is every ML system generative?
2. Is ChatGPT a model name or an application name?
3. Can a token be smaller than a word?
4. Does a larger context window guarantee a more accurate answer?
5. What is the difference between zero-shot and few-shot prompting?
6. If you write “Act as a doctor,” does that guarantee accurate medical advice?
7. Which is safer: fabricating likely causes of lower revenue or listing data required to investigate it?
8. What is one danger of sharing customer data with an AI tool?

## Glossary

| Term | Meaning in one line |
|---|---|
| **LLM** | A model trained to process and generate language/code. |
| **Token** | A unit of input/output text used by a tokenizer. |
| **Tokenizer** | Converts text into the model's token units/IDs. |
| **Embedding** | Numerical representation learned for text or other data. |
| **Transformer** | Neural-network architecture using attention mechanisms. |
| **Attention** | Mechanism for combining information across relevant token positions. |
| **Parameter** | A learned numerical setting within a model. |
| **Context window** | Token capacity of a model request/configuration. |
| **Inference** | Using a trained model to generate a result. |
| **Fine-tuning** | Additional training to adapt a model. |
| **Multimodal** | Working across multiple input/output types (e.g., text and images). |
| **Prompt engineering** | Designing requests and context for useful responses. |
| **Few-shot** | Providing multiple examples as guidance. |
| **Hallucination** | Plausible-sounding but unsupported or false model output. |
| **Open-weight** | Model weights are available for use under specified terms. |
| **Tool use** | An AI system interacts with external capabilities such as search or code execution. |
