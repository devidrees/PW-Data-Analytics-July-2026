````python
# error handling: we use Try Except blocks
print("idrees")

try:
  numbers = [1,2,3,4,5]
  print(numbers[10])
except IndexError:
  print("Error. You are accessing, something that does not exist!")

print("hello")
````

````python
try:
  result = 10/0

except Exception as e:
  print("Error.", e)
````

````python
def calculator(a,b):
  print("Sum", a + b)
  print("Division:", a/b)
  if a > 100:
    raise ValueError("a must be less than 100")
  sum = a + b
  return sum
````

````python
try:
  calculator(34,10) # but it is a string
  result = 10 + 56
except  ZeroDivisionError as e:
  print("Don't Divide with Zero!", e)
except TypeError as e:
  print("Type Mismatch!", e)
except Exception as e: # generic fallback for any error we could not expect
  print("Error has occured!", e)
else:
  # additional behaviour after you have handled errors
  print("result:", result)
finally:
  print("Execution Complete!")
````
