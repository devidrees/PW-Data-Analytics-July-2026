```python
#
# user aapko two numbers
# numbers ka sum, diff, mulipllication, division
a = 45
b = 30

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
# user defined functions
# function definition
# local scope: Variables you define inside a function stay inside the function
# global scope: variables in main program can be used anywhere even inside functions

# a = 10
# b = 20

def idrees_calc(a=1, b=1): # function parameters - placeholders
  # we added default arguments as well
  print("Sum: ", a + b)
  print("Diff: ", a - b)
  print("Mul: ", a * b)
  print("Div: ", a / b)
  print("Raised to power: ", a ** b)
  print("Square root of both numbers: ", a ** (1/2), b ** (1/2))
```

```python
idrees_calc()
```

```python
# function call - jo bhi function dabbe mai hai, usko execute karo

# positional arguments: you pass arguments according to position or order in the function definition
idrees_calc(10,3) # Arguments - actual inputs
```

```python
# Keyword arguments: pass arguments by parameter's name - so order doesn't matter
idrees_calc(b = 20, a = 3)
```
