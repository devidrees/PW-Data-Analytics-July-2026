### Lists vs Arrays

- Memory Efficiency: NumPy stores elements in contiguous memory blocks of a uniform data type, whereas standard Python lists store references/pointers to separate objects in memory.

-

1. Numpy arrays are faster to use than python lists
  - faster to calculate with - accessing elements, operating them
  -
2. Arrays save data while Lists save addresses to the data

3. ARRARYs more efficient in SPEED as well as MEMORY

- if you are using lesser number of elements
- 1000s of elements save kar rahe ho, tab maybe speed mai aapko difference nahi dikhega
- but with huge data differences become stark - let's say more than 10_000 elements

## Aggregation:
- statistical measures
- min max sum mean std var

```python
arr_2d = np.array([
    [ 1,  2,  3],
    [10, 20, 10],
    [20, 20, 20]
    ])

print(arr_2d.sum())

print(arr_2d.sum(axis= 0)) # COLUMN WISE chalo - Go down the ROWS
print(arr_2d.sum(axis= 1)) # ROW WISE chalo - Go across the COLUMNS


print(arr_2d.mean(axis= 0)) #
```

## Matrix operations:
- numpy arrays can **broadcast**

```python
# scalar operation - any number
[1,2,3] *  2
'idrees' * 3
"-" * 10

# matrix operations
[1,2,3] * 2 # [2,4,6]

# arrays can do this: Broadcasting
# ELEMENT WISE OPERATIONS
np.array([1,2,3]) * 2

np.array([1,2,3]) * np.array([10, 20, 30])
```

```python
a = np.array([1,2,3,4])

b = np.array([
      [10,20,30,40],
      [10,20,30,40],
      [10,20,30,40],
      [10,20,30,40]
    ])

a * b

# ONLY WORKS ON COMPATIBLE SHAPES
```
