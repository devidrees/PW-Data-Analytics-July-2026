### Aggregates in numpy

- axis = 0 ->
 - operates columnwise - DOWN THE ROWS
 - aggregate the columns and return result
- axis = 1 ->
  - operates rowwise - ACROSS THE COLUMNS
  - aggregate the rows and return result



```python
arr = np.array([
    [10, 20, 30], # row 1:
    [10, 20, 30], # row 2
    [10, 20, 30]  # row 3
])

grand_mean = arr.mean() # mean of all the elements

# axis = 0 -> aggregate the columns and return result
colwise_mean = arr.mean(axis= 0) # mean of each column
rowwise_mean = arr.mean(axis= 1) # mean of each column

arr.sum()
arr.min()
arr.max()
arr.std()
arr.var()
```

# Handling Missing Values (NaN)


```python
df.isna().sum()   # Counts missing values per column

# Drop any row that has a NaN in any column
df_dropped = df.dropna()

# Drop rows only if 'price' is missing
df_dropped_price = df.dropna(subset=['price'])
```

```python
# Imputation Strategies

df_const = df.copy()

# Fill missing prices with 0
df_const['price'] = df_const['price'].fillna(0)
```

- Mean Imputation:
- When to use: Data has a normal/symmetric distribution with no extreme outliers

```python
df_mean = df.copy()

# Calculate overall mean price (ignoring NaNs automatically)
mean_val = df_mean['price'].mean()
print(f"Global Mean Price: {mean_val:.2f}")

# Fill NaNs
df_mean['price'] = df_mean['price'].fillna(mean_val)

print(df_mean[['order_id', 'category', 'price']].iloc[[3, 7]])
```

- Median Imputation
  - Fill missing entries with the middle value (50th percentile).
  - When to use: Data contains heavy outliers (e.g., a few expensive luxury items that would distort the simple average).

```python
df_median = df.copy()

# Calculate median price
median_val = df_median['price'].median()
print(f"Global Median Price: {median_val:.2f}")

# Fill NaNs
df_median['price'] = df_median['price'].fillna(median_val)

print(df_median[['order_id', 'category', 'price']].iloc[[3, 7]])
```

- Mode Imputation (Most Frequent Value)
  - Fill with the most repeated value in the column.
  - When to use: Categorical columns (like filling a missing city or category), or discrete values with a dominant common price tier.

```python
df_mode = df.copy()

# .mode() returns a Series because there can be ties; take index [0]
mode_val = df_mode['price'].mode()[0]
print(f"Mode Price: {mode_val}")

df_mode['price'] = df_mode['price'].fillna(mode_val)

print(df_mode[['order_id', 'category', 'price']].iloc[[3, 7]])
```

- Forward Fill (ffill) and Backward Fill (bfill)
  - ffill: Carry the previous valid value forward into the missing spot
  - bfill: Pull the next valid value backward into the missing spot
  - When to use: Time-series, chronological order logs, stock prices, or sensor readings where values are expected to persist from the prior reading

```python
df_time_series = df.copy()

# Forward fill: row 3 takes the price from row 2 (600.0)
df_time_series['price_ffill'] = df_time_series['price'].ffill()

# Backward fill: row 3 takes the price from row 4 (2500.0)
df_time_series['price_bfill'] = df_time_series['price'].bfill()

print(df_time_series[['order_id', 'price', 'price_ffill', 'price_bfill']].iloc[[2, 3, 4]])
```

- Group-Conditioned Imputation (Category Mean)
  - Don't apply one global mean across everything
  - Calculate the mean per category
  - (e.g., Home vs. Clothing) and fill each missing item with its own category's average
  

  - When to use: The standard real-world best practice for e-commerce, because an average Home item price shouldn't be skewed by Electronics

```python
df_grouped = df.copy()

# transform('mean') creates a Series matching the original row alignment
category_means = df_grouped.groupby('category')['price'].transform('mean')

# Fill missing values using that category-aligned Series
df_grouped['price'] = df_grouped['price'].fillna(category_means)
df_grouped['total_amount'] = df_grouped['total_amount'].fillna(category_means)

print(df_grouped[['order_id', 'category', 'price']].iloc[[3, 7]])
```

## Subplots
-
 - To create a $2 \times 2$ grid (4 visuals), you first define the layout.
 - Think of it like an Excel sheet where you have to tell the computer exactly which cell (row and column) the chart should live in.

```python
# 1. Create a 1-row, 2-column layout (ax1 is left, ax2 is right)
fig, ax = plt.subplots(1, 2, figsize=(8, 4))

# 2. Draw Left Plot (Sales)
ax[0].bar(df['category'], df['price'], color='steelblue')
ax[0].set_title('Total Sales ($)')

# 3. Draw Right Plot (Quantity)
ax[1].bar(df['category'], df['qty'], color='orange')
ax[1].set_title('Total Items Sold')
```

```python
# 2. CREATE A 2x2 GRID (DASHBOARD CANVAS)
# ---------------------------------------------------------
# fig = the whole window/page
# axes = grid of 4 mini-plots:
#   axes[0, 0] = Top-Left
#   axes[0, 1] = Top-Right
#   axes[1, 0] = Bottom-Left
#   axes[1, 1] = Bottom-Right

fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(12, 8))
fig.suptitle('E-Commerce Performance Dashboard', fontsize=16, fontweight='bold')

# PANEL 1 (Top-Left): Total Revenue by Category (Bar Chart)
# ---------------------------------------------------------
cat_rev = df.groupby('category')['total_amount'].sum()

axes[0, 0].bar(cat_rev.index, cat_rev.values, color='steelblue', edgecolor='black')
axes[0, 0].set_title('Total Revenue by Category')
axes[0, 0].set_xlabel('Category')
axes[0, 0].set_ylabel('Revenue ($)')
axes[0, 0].grid(axis='y', linestyle='--', alpha=0.6)


# PANEL 2 (Top-Right): Customer Spending (Horizontal Bar)
# ---------------------------------------------------------
cust_spend = df.groupby('customer')['total_amount'].sum().sort_values()

axes[0, 1].barh(cust_spend.index, cust_spend.values, color='coral', edgecolor='black')
axes[0, 1].set_title('Total Spend by Customer')
axes[0, 1].set_xlabel('Total Spend ($)')
axes[0, 1].set_ylabel('Customer')
axes[0, 1].grid(axis='x', linestyle='--', alpha=0.6)


# PANEL 3 (Bottom-Left): Unit Price vs. Quantity (Scatter)
# ---------------------------------------------------------
axes[1, 0].scatter(df['price'], df['qty'], color='seagreen', s=80, alpha=0.8, edgecolor='black')
axes[1, 0].set_title('Item Price vs. Quantity Ordered')
axes[1, 0].set_xlabel('Price ($)')
axes[1, 0].set_ylabel('Quantity')
axes[1, 0].grid(True, linestyle=':', alpha=0.6)


# PANEL 4 (Bottom-Right): Revenue by Order Timeline (Line)
# ---------------------------------------------------------
axes[1, 1].plot(df['order_id'], df['total_amount'], marker='o', color='purple', linewidth=2)
axes[1, 1].set_title('Order Value Timeline')
axes[1, 1].set_xlabel('Order ID')
axes[1, 1].set_ylabel('Order Total ($)')
axes[1, 1].set_xticks(df['order_id'])   # Label every single order ID
axes[1, 1].grid(True, linestyle='--', alpha=0.6)

# 3. CLEAN UP SPACING AND DISPLAY
# ---------------------------------------------------------
# tight_layout stops titles and axis labels from overlapping
plt.tight_layout()
plt.show()
```

### Dashboards with simple prompts

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns

# ------------------------------------------------------------------
# 1. DATASET
# ------------------------------------------------------------------
data = {
    'order_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'customer': ['Aarav', 'Pooja', 'Rahul', 'Sneha', 'Aarav', 'Vikram', 'Pooja', 'Zayd', 'Sneha', 'Aarav'],
    'category': ['Electronics', 'Clothing', 'Clothing', 'Home', 'Electronics', 'Home', 'Electronics', 'Clothing', 'Electronics', 'Home'],
    'price': [1200.0, 450.0, 600.0, 500.0, 2500.0, 300.0, 800.0, 525.0, 150.0, 700.0],
    'qty': [1, 2, 1, 3, 1, 4, 2, 2, 5, 2],
    'total_amount': [1200.0, 900.0, 600.0, 500.0, 2500.0, 1200.0, 1600.0, 525.0, 750.0, 1400.0]
}
df = pd.DataFrame(data)

# ------------------------------------------------------------------
# 2. POWER BI THEME CONFIGURATION
# ------------------------------------------------------------------
BG_COLOR = "#F4F6F9"      # Modern light grey canvas
CARD_BG = "#FFFFFF"       # White card background
PRIMARY = "#1E40AF"       # Corporate deep blue
SECONDARY = "#0D9488"     # Teal
ACCENT = "#F59E0B"        # Amber
TEXT_DARK = "#1E293B"     # Slate dark text
TEXT_MUTED = "#64748B"    # Grey text for subtitles/labels

plt.rcParams.update({
    'font.family': 'sans-serif',
    'axes.edgecolor': '#E2E8F0',
    'axes.linewidth': 0.8,
    'grid.color': '#E2E8F0',
    'grid.linestyle': '--',
    'grid.alpha': 0.7
})

# ------------------------------------------------------------------
# 3. DASHBOARD CANVAS & LAYOUT (GridSpec)
# ------------------------------------------------------------------
fig = plt.figure(figsize=(15, 9), facecolor=BG_COLOR)
# 3 rows: Row 0 = KPIs, Row 1 & 2 = Visualizations
gs = gridspec.GridSpec(nrows=3, ncols=4, height_ratios=[0.8, 2.2, 2.2], hspace=0.35, wspace=0.25)

# Dashboard Master Header
fig.text(0.05, 0.96, "E-COMMERCE EXECUTIVE PERFORMANCE DASHBOARD",
         fontsize=16, fontweight='bold', color=TEXT_DARK)
fig.text(0.05, 0.935, "Overview of sales revenue, top customers, and order distributions",
         fontsize=10, color=TEXT_MUTED)

# ------------------------------------------------------------------
# 4. KPI CARDS (ROW 0)
# ------------------------------------------------------------------
kpis = [
    ("TOTAL REVENUE", f"${df['total_amount'].sum():,.0f}", PRIMARY),
    ("TOTAL ORDERS", f"{df['order_id'].nunique()}", SECONDARY),
    ("UNITS SOLD", f"{df['qty'].sum()} pcs", ACCENT),
    ("AVG ORDER VALUE", f"${df['total_amount'].mean():,.1f}", "#6366F1")
]

for col_idx, (title, value, color) in enumerate(kpis):
    ax_kpi = fig.add_subplot(gs[0, col_idx])
    ax_kpi.set_facecolor(CARD_BG)
    ax_kpi.set_xticks([])
    ax_kpi.set_yticks([])
    for spine in ax_kpi.spines.values():
        spine.set_color('#CBD5E1')
        spine.set_linewidth(1)

    # Card Title & Metric Value
    ax_kpi.text(0.5, 0.68, title, fontsize=9, fontweight='semibold',
                color=TEXT_MUTED, ha='center', va='center')
    ax_kpi.text(0.5, 0.32, value, fontsize=18, fontweight='bold',
                color=color, ha='center', va='center')


# ------------------------------------------------------------------
# 5. CHART 1: REVENUE BY CATEGORY (Row 1, Left)
# ------------------------------------------------------------------
ax1 = fig.add_subplot(gs[1, 0:2])
ax1.set_facecolor(CARD_BG)

cat_data = df.groupby('category')['total_amount'].sum().reset_index()
cat_data = cat_data.sort_values(by='total_amount', ascending=False)

bars = ax1.bar(cat_data['category'], cat_data['total_amount'], color=PRIMARY, width=0.55)
ax1.set_title("Total Revenue by Category", fontsize=11, fontweight='bold', color=TEXT_DARK, pad=12)
ax1.set_ylabel("Revenue ($)", fontsize=9, color=TEXT_MUTED)
ax1.grid(axis='y')

# Data Labels on top of bars
for bar in bars:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2, height + 100, f"${height:,.0f}",
             ha='center', va='bottom', fontsize=9, fontweight='semibold', color=TEXT_DARK)

ax1.set_ylim(0, cat_data['total_amount'].max() * 1.15)


# ------------------------------------------------------------------
# 6. CHART 2: CUSTOMER SPENDING RANKING (Row 1, Right)
# ------------------------------------------------------------------
ax2 = fig.add_subplot(gs[1, 2:4])
ax2.set_facecolor(CARD_BG)

cust_data = df.groupby('customer')['total_amount'].sum().sort_values(ascending=True)

bars_h = ax2.barh(cust_data.index, cust_data.values, color=SECONDARY, height=0.55)
ax2.set_title("Customer Contribution Ranking", fontsize=11, fontweight='bold', color=TEXT_DARK, pad=12)
ax2.set_xlabel("Total Spend ($)", fontsize=9, color=TEXT_MUTED)
ax2.grid(axis='x')

# Value labels beside horizontal bars
for bar in bars_h:
    width = bar.get_width()
    ax2.text(width + 100, bar.get_y() + bar.get_height()/2, f"${width:,.0f}",
             ha='left', va='center', fontsize=9, fontweight='semibold', color=TEXT_DARK)

ax2.set_xlim(0, cust_data.max() * 1.2)


# ------------------------------------------------------------------
# 7. CHART 3: ORDER TIMELINE & BENCHMARK (Row 2, Left)
# ------------------------------------------------------------------
ax3 = fig.add_subplot(gs[2, 0:2])
ax3.set_facecolor(CARD_BG)

ax3.plot(df['order_id'], df['total_amount'], marker='o', markersize=6,
         color='#4338CA', linewidth=2.2, label='Order Amount')

# Benchmark average reference line
avg_rev = df['total_amount'].mean()
ax3.axhline(avg_rev, color='#EF4444', linestyle=':', linewidth=1.5, label=f'Avg Order (${avg_rev:.0f})')

ax3.set_title("Order Value Trend vs. Average", fontsize=11, fontweight='bold', color=TEXT_DARK, pad=12)
ax3.set_xlabel("Order ID", fontsize=9, color=TEXT_MUTED)
ax3.set_ylabel("Order Value ($)", fontsize=9, color=TEXT_MUTED)
ax3.set_xticks(df['order_id'])
ax3.grid(True)
ax3.legend(frameon=True, facecolor=CARD_BG, edgecolor='#E2E8F0', fontsize=8)


# ------------------------------------------------------------------
# 8. CHART 4: UNIT PRICE VS. QUANTITY (Row 2, Right)
# ------------------------------------------------------------------
ax4 = fig.add_subplot(gs[2, 2:4])
ax4.set_facecolor(CARD_BG)

scatter = ax4.scatter(
    df['price'],
    df['qty'],
    s=df['total_amount'] / 6,   # Bubble size represents total transaction value
    c=df['total_amount'],       # Color gradient based on transaction value
    cmap='Blues',
    edgecolor='#0F172A',
    linewidth=0.8,
    alpha=0.85
)

ax4.set_title("Price vs. Quantity (Bubble Size = Total Bill)", fontsize=11, fontweight='bold', color=TEXT_DARK, pad=12)
ax4.set_xlabel("Unit Price ($)", fontsize=9, color=TEXT_MUTED)
ax4.set_ylabel("Quantity Purchased", fontsize=9, color=TEXT_MUTED)
ax4.grid(True)

# Mini colorbar on the side
cbar = fig.colorbar(scatter, ax=ax4, orientation='vertical', shrink=0.7, pad=0.03)
cbar.set_label("Total Bill ($)", fontsize=8, color=TEXT_MUTED)
cbar.ax.tick_params(labelsize=8)

# ------------------------------------------------------------------
# 9. RENDER
# ------------------------------------------------------------------
plt.tight_layout()
plt.savefig('ecommerce_dashboard.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ------------------------------------------------------------------
# 1. SYNTHETIC DATASET (unchanged)
# ------------------------------------------------------------------
np.random.seed(42)
categories = ['Electronics', 'Home & Living', 'Apparel', 'Accessories']
customers = ['Aarav', 'Pooja', 'Rahul', 'Sneha', 'Vikram', 'Zayd', 'Meera', 'Rohan']
cities = ['Mumbai', 'Delhi', 'Bengaluru', 'Srinagar', 'Hyderabad', 'Pune']

df = pd.DataFrame({
    'order_id': [1000 + i for i in range(60)],
    'customer': np.random.choice(customers, 60),
    'category': np.random.choice(categories, 60, p=[0.35, 0.25, 0.25, 0.15]),
    'city': np.random.choice(cities, 60),
    'price': np.random.choice([150, 300, 450, 600, 800, 1200, 2200, 3100], 60),
    'qty': np.random.randint(1, 5, 60),
    'rating': np.random.uniform(3.8, 5.0, 60).round(1)
})
df['revenue'] = df['price'] * df['qty']
df['margin'] = (df['revenue'] * np.random.uniform(0.20, 0.40, 60)).round(2)
df = df.sort_values(by='order_id').reset_index(drop=True)

CUR = "₹"   # change to "$" if you prefer

# ------------------------------------------------------------------
# 2. PALETTE
# ------------------------------------------------------------------
BG_MAIN, CARD_BG, BORDER = '#080C14', '#111827', '#1E293B'
CYAN, TEAL, AMBER, PURPLE = '#38BDF8', '#2DD4BF', '#FBBF24', '#A78BFA'
TEXT_PRIMARY, TEXT_MUTED = '#F8FAFC', '#94A3B8'
FONT = "Inter, -apple-system, BlinkMacSystemFont, 'Segoe UI', sans-serif"

cat_palette = {'Electronics': CYAN, 'Home & Living': TEAL,
               'Apparel': PURPLE, 'Accessories': AMBER}

selections = [None] + categories          # slicer states
sub_of = lambda s: df if s is None else df[df.category == s]

# ------------------------------------------------------------------
# 3. GRID (explicit row heights -> KPI cards line up exactly)
# ------------------------------------------------------------------
HEIGHT, MT, MB = 1020, 150, 50
PLOT_H = HEIGHT - MT - MB

fig = make_subplots(
    rows=3, cols=4,
    specs=[
        [{"type": "indicator"}] * 4,
        [{"colspan": 2, "type": "xy"}, None, {"colspan": 2, "type": "xy"}, None],
        [{"colspan": 2, "type": "xy"}, None, {"colspan": 2, "type": "xy"}, None],
    ],
    row_heights=[0.15, 0.425, 0.425],
    vertical_spacing=0.095,
    horizontal_spacing=0.05,
    subplot_titles=("", "", "", "",
                    "REVENUE BY CATEGORY (SELECTION HIGHLIGHTED)",
                    "PRICE VS ORDER VALUE (BUBBLE = QTY)",
                    "CUMULATIVE REVENUE BY ORDER",
                    "ORDER VALUE DISTRIBUTION"),
)

# ------------------------------------------------------------------
# 4. KPI INDICATORS: one set per slicer state, so KPIs react to filter
# ------------------------------------------------------------------
PRIOR = [0.88, 0.90, 0.94, 0.92]   # illustrative prior-period factors


def kpi_values(sub):
    rev = sub.revenue.sum()
    return [rev, len(sub), sub.revenue.mean(), sub.margin.sum() / rev * 100]


kpi_meta = [  # title, prefix, suffix, number format, colour
    ("TOTAL REVENUE", CUR, "", ",.0f", CYAN),
    ("TOTAL ORDERS", "", "", ",.0f", TEAL),
    ("AVG BASKET SIZE", CUR, "", ",.0f", AMBER),
    ("GROSS MARGIN", "", "%", ".1f", PURPLE),
]

for s in selections:
    vals = kpi_values(sub_of(s))
    for k, (title, pre, suf, fmt, color) in enumerate(kpi_meta):
        fig.add_trace(go.Indicator(
            mode="number+delta",
            value=vals[k],
            visible=(s is None),
            delta=dict(reference=vals[k] * PRIOR[k], relative=True, valueformat=".1%",
                       increasing=dict(color="#34D399"), decreasing=dict(color="#F87171"),
                       font=dict(size=12, family=FONT)),
            number=dict(prefix=pre, suffix=suf, valueformat=fmt,
                        font=dict(size=32, color=color, family=FONT)),
            title=dict(text=f"<b>{title}</b>",
                       font=dict(size=11, color=TEXT_MUTED, family=FONT)),
        ), row=1, col=k + 1)

# ------------------------------------------------------------------
# 5. CHARTS
# ------------------------------------------------------------------
# 5a. Bar: all categories always visible, selected one highlighted
all_rev = [df[df.category == c].revenue.sum() for c in categories]
for s in selections:
    fig.add_trace(go.Bar(
        x=categories, y=all_rev,
        marker=dict(color=[cat_palette[c] for c in categories],
                    opacity=[1 if (s is None or c == s) else 0.22 for c in categories],
                    cornerradius=6),
        text=[f"{CUR}{v/1000:,.1f}K" for v in all_rev],
        textposition="outside", cliponaxis=False,
        textfont=dict(color=TEXT_PRIMARY, size=11),
        hovertemplate="<b>%{x}</b><br>Revenue: <b>" + CUR + "%{y:,.0f}</b><extra></extra>",
        visible=(s is None), showlegend=False,
    ), row=2, col=1)

# 5b. Bubble scatter
for cat in categories:
    sub = df[df.category == cat]
    fig.add_trace(go.Scatter(
        x=sub.price, y=sub.revenue, mode='markers', name=cat,
        marker=dict(size=sub.qty * 7, color=cat_palette[cat], opacity=0.8,
                    line=dict(width=0.8, color='#FFFFFF')),
        text=sub.customer, customdata=sub.qty,
        hovertemplate="<b>%{text}</b> (" + cat + ")<br>Unit price: <b>" + CUR +
                      "%{x:,}</b><br>Order value: <b>" + CUR +
                      "%{y:,}</b><br>Qty: <b>%{customdata}</b><extra></extra>",
        showlegend=False,
    ), row=2, col=3)

# 5c. Cumulative revenue (clean step lines instead of zig-zag spaghetti)
for cat in categories:
    sub = df[df.category == cat]
    fig.add_trace(go.Scatter(
        x=sub.order_id, y=sub.revenue.cumsum(), mode='lines', name=cat,
        line=dict(color=cat_palette[cat], width=2.5, shape='hv'),
        hovertemplate="Order #%{x}<br>Cumulative: <b>" + CUR + "%{y:,.0f}</b><extra></extra>",
        showlegend=False,
    ), row=3, col=1)

# 5d. Box plots
for cat in categories:
    sub = df[df.category == cat]
    fig.add_trace(go.Box(
        y=sub.revenue, name=cat, marker_color=cat_palette[cat],
        line=dict(width=1.5), boxmean=True, boxpoints='outliers',
        hovertemplate="<b>" + cat + "</b><br>" + CUR + "%{y:,.0f}<extra></extra>",
        showlegend=False,
    ), row=3, col=3)

# ------------------------------------------------------------------
# 6. SLICER (trace order: 20 KPI, 5 bar, 4 scatter, 4 line, 4 box)
# ------------------------------------------------------------------
def visibility(sel_idx):
    s = selections[sel_idx]
    m = []
    for i in range(len(selections)):
        m += [i == sel_idx] * 4                       # KPI sets
    m += [i == sel_idx for i in range(len(selections))]  # bar sets
    for _ in range(3):                                 # scatter, line, box
        m += [(s is None or c == s) for c in categories]
    return m


def header(sel):
    sub = ("Cross-departmental performance, volume and margin" if sel is None
           else f"Filtered to {sel} · KPIs and charts update together")
    return (f"<b>EXECUTIVE REVENUE DASHBOARD</b><br>"
            f"<span style='font-size:11px;color:{TEXT_MUTED};'>{sub}</span>")


buttons = [dict(label=("ALL" if s is None else s.upper()), method="update",
                args=[{"visible": visibility(i)}, {"title.text": header(s)}])
           for i, s in enumerate(selections)]

# ------------------------------------------------------------------
# 7. KPI CARDS: positioned from the REAL indicator domains
# ------------------------------------------------------------------
shapes = []
for k in range(4):
    dx, dy = fig.data[k].domain.x, fig.data[k].domain.y
    shapes.append(dict(type="rect", xref="paper", yref="paper",
                       x0=dx[0], x1=dx[1], y0=dy[0], y1=dy[1],
                       fillcolor=CARD_BG, line=dict(color=BORDER, width=1), layer="below"))
    shapes.append(dict(type="rect", xref="paper", yref="paper",           # accent strip
                       x0=dx[0], x1=dx[1], y0=dy[1] - 0.006, y1=dy[1],
                       fillcolor=kpi_meta[k][4], line=dict(width=0), layer="below"))

# ------------------------------------------------------------------
# 8. LAYOUT
# ------------------------------------------------------------------
fig.update_layout(
    height=HEIGHT, autosize=True,
    paper_bgcolor=BG_MAIN, plot_bgcolor=CARD_BG,
    showlegend=False, shapes=shapes,
    font=dict(family=FONT, color=TEXT_PRIMARY),
    title=dict(text=header(None), x=0.0, xanchor="left", y=0.97, yanchor="top",
               font=dict(size=18, color=TEXT_PRIMARY)),
    updatemenus=[dict(
        type="buttons", direction="right", active=0, showactive=True,
        x=1.0, xanchor="right", y=1 + 118 / PLOT_H, yanchor="top",  # paper coords: sits in the header band, above the KPI cards
        bgcolor="#1E3A8A", bordercolor="#3B82F6", borderwidth=1,
        font=dict(size=10, color="#FFFFFF", family=FONT),
        pad=dict(r=2, t=2, b=2, l=2), buttons=buttons)],
    hoverlabel=dict(bgcolor='#0F172A', font_size=12, font_family=FONT,
                    font_color=TEXT_PRIMARY, bordercolor='#334155'),
    margin=dict(l=45, r=45, t=MT, b=MB),
    bargap=0.45,
)

# Chart panels: bordered boxes, readable ticks, currency formatting
axis_common = dict(showgrid=True, gridcolor=BORDER, zeroline=False,
                   showline=True, linecolor=BORDER, mirror=True,
                   tickfont=dict(size=10, color=TEXT_MUTED, family=FONT),
                   title_font=dict(size=10, color=TEXT_MUTED), automargin=True)
fig.update_xaxes(**axis_common)
fig.update_yaxes(**axis_common)

fig.update_yaxes(tickprefix=CUR, tickformat=",.0f")
fig.update_yaxes(range=[0, max(all_rev) * 1.18], title_text="Revenue", row=2, col=1)
fig.update_xaxes(showgrid=False, row=2, col=1)
fig.update_xaxes(tickprefix=CUR, tickformat=",.0f", title_text="Unit price", row=2, col=3)
fig.update_yaxes(title_text="Order value", row=2, col=3)
fig.update_xaxes(title_text="Order ID", row=3, col=1)
fig.update_yaxes(title_text="Cumulative revenue", row=3, col=1)
fig.update_xaxes(showgrid=False, row=3, col=3)
fig.update_yaxes(title_text="Order value", row=3, col=3)

# Subplot titles: left-aligned to each panel, bold, consistent
axes = [fig.layout.xaxis, fig.layout.xaxis2, fig.layout.xaxis3, fig.layout.xaxis4]
chart_titles = [a for a in fig.layout.annotations if a.text]
for ann, ax in zip(chart_titles, axes):
    ann.update(text=f"<b>{ann.text}</b>", x=ax.domain[0], xanchor="left", yanchor="bottom",
               yshift=6, font=dict(size=11, color=TEXT_PRIMARY, family=FONT))

# ------------------------------------------------------------------
# 9. RENDER
# ------------------------------------------------------------------
fig.write_html("executive_cockpit_clean.html",
               config={"displayModeBar": False, "responsive": True})
fig.show()
```
