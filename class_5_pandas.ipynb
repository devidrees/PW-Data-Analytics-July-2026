#### 1. .loc - Location by label

#### 2. .iloc - integer location
- we use integers to locate things
- [row number, column number]

```python
titanic_df.loc[2] # a row label
titanic_df.loc[2, 'age'] # row label, column label

titanic_df.loc[[0,1,2,3], ['survived',	'pclass',	'sex','age']]
titanic_df.loc[0:4, ['survived',	'pclass',	'sex','age']]
```

## Grouping and Aggregation
-




```python
import pandas as pd

data = {
    "Department": ["Sales", "Sales", "IT", "IT", "HR", "HR", "Sales"],
    "City": [
        "New York",
        "Chicago",
        "New York",
        "Chicago",
        "New York",
        "Chicago",
        "New York",
    ],
    "Salary": [70000, 65000, 95000, 88000, 60000, 58000, 72000],
    "Bonus": [5000, 4500, 10000, 7000, 3000, 2500, 6000],
    "Experience": [3, 2, 6, 5, 4, 2, 4],
}

df = pd.DataFrame(data)
```

```python
# Aggregation in pandas

df.groupby(["Department"])["Salary"].sum()
df.groupby(["Department"])["Salary"].mean()


df.groupby(["Department"])["Salary"].agg(["mean","sum","min","count"])

df.groupby(["Department","City"])["Salary"].sum().sort_values(ascending= True)
```
