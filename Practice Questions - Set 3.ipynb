Problem 2: Given a 2D array of student marks across different subjects, write code to calculate the average mark for each student along the rows and the maximum mark for each subject down the columns.

Problem 4: Given two (2, 2) matrices A and B, compute both their element-wise product A * B and their matrix product A @ B, and print both results to show the difference.

Problem 5: Take a matrix X of shape (2, 3) and a matrix W of shape (3, 2). Matrix multiply them. Then add a 1D matrix b of length 2 to the result. Return the result as y.
