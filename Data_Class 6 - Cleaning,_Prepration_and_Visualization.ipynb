#### Outlier detection
-
- we have visualization
  - box plot
  

```python
ages = [23, 25, 46, 34, 65, 26, 38, 34, 33]

plt.figure(figsize=(5,3))
sns.boxplot(x = ages)

plt.show()
```

```python
from os import name
# We use the IQR method

ages = [23, 25, 46, 34, 65, 26, 38, 34, 33]

df_ages = pd.Series(ages, name= 'ages')

Q1 = df_ages.quantile(0.25)
Q3 = df_ages.quantile(0.75)

IQR = Q3 - Q1

upper_limit = Q3 + IQR*1.5
lower_limit = Q1 - IQR*1.5


df_ages_nooutliers = df_ages[(df_ages > lower_limit) & (df_ages < upper_limit)]
print(df_ages_nooutliers)
```
